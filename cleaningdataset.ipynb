```python
import pandas as pd

df=pd.read_excel("Customer Call List.xlsx")
df=df.drop_duplicates()
df
```

```python
df=df.drop(columns="Not_Useful_Column")
```

```python
df
```

```python
df["Last_Name"]=df["Last_Name"].str.strip("/._123")
```

```python
df
```

```python
df["Phone_Number"]=df["Phone_Number"].str.replace('[^a-zA-Z0-9]','',regex=True)
```

```python
df["Phone_Number"]=df["Phone_Number"].apply(lambda x:str(x))
df["Phone_Number"]=df["Phone_Number"].apply(lambda x:x[0:3]+'-'+x[3:6]+'-'+x[6:10])
```

```python
df["Phone_Number"]=df["Phone_Number"].replace('nan--','')

df['Phone_Number']=df['Phone_Number'].replace("Na--",'')
df
```

```python
df
```

```python
df[["street_Address","State","Zipcode"]]=df["Address"].str.split(',',n=2,expand=True)
df
```

```python
df["Paying Customer"]=df["Paying Customer"].replace('Y','Yes')
df["Paying Customer"]=df['Paying Customer'].replace('N','No')
df
```

```python
df['Do_Not_Contact']=df['Do_Not_Contact'].replace('Y','Yes')
df['Do_Not_Contact']=df['Do_Not_Contact'].replace('N','No')
df
```

```python
df=df.replace('N/a','')
df=df.fillna('')
df
```

```python
for x in df.index:
  if df.loc[x,"Do_Not_Contact"]=="Yes":
    df.drop(x,inplace=True)
df
```

```python
df['Do_Not_Contact']=df['Do_Not_Contact'].replace('',"No")
df
```

```python
df.drop(columns="Address",inplace=True)
df
```

```python
for i in df.index:
  if df.loc[i,"Phone_Number"]=='':
    df.drop(i,inplace=True)
df
```

```python
df=df.reset_index(drop=True)
```
